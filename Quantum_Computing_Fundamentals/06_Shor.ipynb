# 1. The Integer Factorization Problem

Suppose we are given the number

$$
15
$$

The goal is to determine its prime factors.

The answer is

$$
15 = 3 \times 5
$$

For small numbers this is easy.

For extremely large numbers (hundreds or thousands of digits), factoring becomes computationally expensive for classical computers.

Many modern cryptographic systems rely on this difficulty.

Shor's Algorithm provides a quantum method for solving this problem much more efficiently.

## Why Doesn't Shor Factor Directly?

One of the clever ideas behind Shor's Algorithm is that it does **not** search for the factors directly.

Instead, it transforms the factoring problem into a different problem:

**Find the period of a mathematical function.**

Once the period is known, classical mathematics can often recover the factors.

This means the quantum computer solves the difficult part:

- Period Finding

while the classical computer performs the final arithmetic.

# 2. Period Finding

The quantum part of Shor's Algorithm is **not** trying to find the factors directly.

Instead, it finds the **period** of a repeating function.

A **period** is the number of steps required before a sequence begins repeating.

For example, consider the function

$$
3^x \bmod 5
$$

As we increase the value of \(x\), we obtain

| \(x\) | \(3^x mod 5\) |
|------:|----------------:|
| 0 | 1 |
| 1 | 3 |
| 2 | 4 |
| 3 | 2 |
| 4 | 1 |
| 5 | 3 |
| 6 | 4 |
| 7 | 2 |

Notice that the sequence

```text
1, 3, 4, 2
```

repeats over and over again.

The repeating length is called the **period**.

## Finding the Period

The repeating sequence is

```text
1 → 3 → 4 → 2
```

After four steps, the sequence begins again.

Therefore,

$$
r = 4
$$

where \(r\) is called the **period**.

Notice that we are **not** interested in the values themselves.

We are interested in **how often they repeat**.

## Why Does the Period Matter?

Once the period \(r\) has been found, the remaining work is mostly classical mathematics.

For example,

Suppose

- \(N = 15\)
- Choose \(a = 2\)

If the quantum computer finds

$$
r = 4
$$

then

$$
a^{r/2}=2^2=4
$$

Now compute

$$
\gcd(4-1,15)=\gcd(3,15)=3
$$

and

$$
\gcd(4+1,15)=\gcd(5,15)=5
$$

We have recovered the factors

$$
15 = 3 \times 5.
$$

The quantum computer only found the **period**.

The factors are recovered afterwards using classical mathematics.

# 3. Where Does the Quantum Fourier Transform Fit?

The Quantum Fourier Transform (QFT) is the heart of Shor's Algorithm.

Remember:

The goal is **not** to factor a number directly.

The goal is to discover the **period** of a repeating function.

The QFT makes this possible by transforming hidden phase information into measurable information.

Once the period has been found, the remaining calculations are performed classically.

## The Overall Flow of Shor's Algorithm

The complete algorithm can be summarized as:

```text
Choose a Number N

↓

Choose a Value a

↓

Create the Quantum Circuit

↓

Modular Exponentiation

↓

Quantum Fourier Transform (QFT)

↓

Measure

↓

Determine the Period r

↓

Classical Mathematics (GCD)

↓

Recover the Factors
```

Notice that only the middle portion is performed on the quantum computer.

The beginning and end are classical computations.

## Why Does the QFT Find the Period?

Suppose a function repeats every

$$
r
$$

steps.

Instead of examining each repetition one at a time,

the QFT causes all repeating parts of the quantum state to interfere constructively.

The measurement is therefore much more likely to reveal information about the period.

In other words,

the QFT converts

```text
Hidden Repetition

↓

Hidden Phase

↓

Constructive Interference

↓

Measurable Information
```

This is why the QFT is such a powerful tool for period finding.

### Notes

One of the biggest insights while learning Shor's Algorithm was realizing that the Quantum Fourier Transform is **not** searching for factors.

Instead,

- Modular exponentiation creates a repeating structure.
- The QFT identifies the period of that structure.
- Classical mathematics converts the period into the factors.

Thinking of the QFT as a **period finder** rather than a **factoring algorithm** makes the entire algorithm much easier to understand.

````python
from math import gcd

# Number to factor
N = 15

# Choose a value that is coprime with N
a = 2

# Suppose the quantum computer found the period
r = 4

# Recover the factors classically
factor1 = gcd(pow(a, r // 2) - 1, N)
factor2 = gcd(pow(a, r // 2) + 1, N)

print("Factors:", factor1, factor2)
````

# 4. Understanding the Code

Although this code looks simple, it represents the **classical post-processing** stage of Shor's Algorithm.

The quantum computer has already completed its task by finding the period

$$
r = 4.
$$

Everything below is classical mathematics.

## Breaking Down Each Line

```python
N = 15
```

The integer we want to factor.

```python
a = 2
```

Choose a number that is coprime with \(N\).

For this example,

$$
\gcd(2,15)=1
$$

so 2 is a valid choice.

```python
r = 4
```

This is the period discovered by the **quantum computer**.

The classical code assumes the period has already been found.

```python
pow(a, r // 2)
```

computes

$$
a^{r/2}.
$$

Since

```python
r // 2
```

is integer division,

$$
4 // 2 = 2
$$

so

$$
2^2 = 4.
$$

## Why Use `pow()`?

Python's built-in

```python
pow(base, exponent)
```

computes

$$
base^{exponent}.
$$

For example,

```python
pow(2,3)
```

returns

```text
8
```

So,

```python
pow(a, r // 2)
```

means

$$
a^{r/2}.
$$

It is **not**

$$
a^{r^2}
$$

and it is **not**

$$
a^{r^N}.
$$

It simply raises

\(a\)

to the power

\(r/2\).

## Recovering the Factors

Once

$$
a^{r/2}
$$

has been computed,

Shor's Algorithm calculates

$$
\gcd(a^{r/2}-1,N)
$$

and

$$
\gcd(a^{r/2}+1,N).
$$

For this example,

$$
2^2=4
$$

so

$$
\gcd(4-1,15)=3
$$

and

$$
\gcd(4+1,15)=5.
$$

Therefore,

$$
15 = 3 \times 5.
$$

The quantum computer discovered the **period**.

The classical computer recovered the **factors**.

### Notes

One of the easiest misconceptions is to think that the quantum computer directly outputs the factors.

It does not.

The quantum computer outputs information about the **period** of a repeating function.

The classical computer then uses that period to compute the factors using the greatest common divisor (GCD).

This division of labor is what makes Shor's Algorithm a hybrid quantum-classical algorithm.
